```python
# Convert gender into numbers
gender_mapping = {
    "Male": 1,
    "Female": 2,
    "Other": 3
}

df["gender"] = df["gender"].map(gender_mapping)


# Convert smoking history into numbers
smoking_mapping = {
    "never": 1,
    "No Info": 2,
    "current": 3,
    "former": 4,
    "ever": 5,
    "not current": 6
}

df["smoking_history"] = df["smoking_history"].map(smoking_mapping)


# Display the updated dataset
df.head()
```

```python
# Find unique values in gender
print("Unique gender values:")
print(df["gender"].unique())

# Find unique values in smoking history
print("\nUnique smoking history values:")
print(df["smoking_history"].unique())
```
